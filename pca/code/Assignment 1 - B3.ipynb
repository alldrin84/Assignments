# 2D ARRAY

```python
import numpy as np

arr = np.array([
    [10, 11, 12, 13, 14, 15],
    [20, 21, 22, 23, 24, 25],
    [30, 31, 32, 33, 34, 35],
    [40, 41, 42, 43, 44, 45],
    [50, 51, 52, 53, 54, 55],
    [60, 61, 62, 63, 64, 65]
])

print(arr.shape)
```

```python
# # 2D Array Indexing & Slicing

# Extract the following using a single indexing/slicing statement
# 1. Every alternate row

# Expected Output:

# [
#  [10,11,12,13,14,15],
#  [30,31,32,33,34,35],
#  [50,51,52,53,54,55]
# ]
```

```python
print(arr[0::2,0::])
```

```python
# # 2. Every alternate column

# Expected Output:

# [
#  [10,12,14],
#  [20,22,24],
#  [30,32,34],
#  [40,42,44],
#  [50,52,54],
#  [60,62,64]
# ]
```

```python
print(arr[0::,0:5:2])
```

```python
# 3. First column

# Expected Output:

# [10,20,30,40,50,60]
```

```python
print(arr[0:6:1,0:2:4])
```

```python
# 4. Last column

# Expected Output:

# [15,25,35,45,55,65]
```

```python
print(arr[0:6:1,5:6:1])
```

```python
# 5. Middle row

# Expected Output:

# [40,41,42,43,44,45]
```

```python
print(arr[3:4:3,::])
```

```python
# 6. Middle column

# Expected Output:

# [13,23,33,43,53,63]
```

```python
print(arr[::,3:4:3])
```

```python
# 7. First three rows

# Expected Output:

# [
#  [10,11,12,13,14,15],
#  [20,21,22,23,24,25],
#  [30,31,32,33,34,35]
# ]
```

```python
print(arr[0:3:,::])
```

```python
# 8. Last three rows

# Expected Output:

# [
#  [40,41,42,43,44,45],
#  [50,51,52,53,54,55],
#  [60,61,62,63,64,65]
# ]
```

```python
print(arr[3:6:,::])
```

```python
# 9. First three columns

# Expected Output:

# [
#  [10,11,12],
#  [20,21,22],
#  [30,31,32],
#  [40,41,42],
#  [50,51,52],
#  [60,61,62]
# ]

```

```python
print(arr[::,0:3:])
```

```python
# 10. Last three columns

# Expected Output:

# [
#  [13,14,15],
#  [23,24,25],
#  [33,34,35],
#  [43,44,45],
#  [53,54,55],
#  [63,64,65]
# ]
```

```python
print(arr[::,3:6:])
```

```python
# 11. Reverse all rows

# Expected Output:

# [
#  [60,61,62,63,64,65],
#  [50,51,52,53,54,55],
#  [40,41,42,43,44,45],
#  [30,31,32,33,34,35],
#  [20,21,22,23,24,25],
#  [10,11,12,13,14,15]
# ]
```

```python
print(arr[7::-1,::])
```

```python
# 12. Reverse all columns

# Expected Output:

# [
#  [15,14,13,12,11,10],
#  [25,24,23,22,21,20],
#  [35,34,33,32,31,30],
#  [45,44,43,42,41,40],
#  [55,54,53,52,51,50],
#  [65,64,63,62,61,60]
# ]
```

```python
print(arr[::,6::-1])
```

```python
# 13. Reverse rows and columns simultaneously

# Expected Output:


# [
#  [65,64,63,62,61,60],
#  [55,54,53,52,51,50],
#  [45,44,43,42,41,40],
#  [35,34,33,32,31,30],
#  [25,24,23,22,21,20],
#  [15,14,13,12,11,10]
# ]
```

```python
print(arr[6::-1,6::-1])
```
